10. Write a Python program to merge two dictionaries.

```python
student1 = {
    "name": "Vaishnavi",
    "age": 22
}

student2 = {
    "branch": "ECE",
    "college": "GNDEC"
}

student = {**student1,**student2}

print(student)
```

You can also merge dictionaries using the `.update()` method. This method adds key-value pairs from one dictionary to another. If a key already exists, its value will be updated.

```python
student1 = {
    "name": "Vaishnavi",
    "age": 22
}

student2 = {
    "branch": "ECE",
    "college": "GNDEC"
}

student1.update(student2)

print(student1)
```

For Python 3.9+, you can also use the `|` operator to merge dictionaries. This creates a *new* dictionary from the two merged dictionaries.

```python
student1 = {
    "name": "Vaishnavi",
    "age": 22
}

student2 = {
    "branch": "ECE",
    "college": "GNDEC"
}

# Using the | operator to merge dictionaries (Python 3.9+)
student = student1 | student2

print(student)
```

You can merge multiple dictionaries at once using the `**` unpacking operator (for older Python versions) or the `|` operator (Python 3.9+). Both methods create a new dictionary from the merged dictionaries.

```python
dict1 = {'a': 1, 'b': 2}
dict2 = {'c': 3, 'd': 4}
dict3 = {'e': 5, 'f': 6}

# Using the ** unpacking operator
merged_dict_unpack = {**dict1, **dict2, **dict3}
print(f"Merged using **: {merged_dict_unpack}")

# Using the | operator (Python 3.9+)
merged_dict_pipe = dict1 | dict2 | dict3
print(f"Merged using |: {merged_dict_pipe}")

# Example with overlapping keys (rightmost dict's value wins)
dict_overlap_1 = {'a': 1, 'b': 2}
dict_overlap_2 = {'b': 3, 'c': 4}
dict_overlap_3 = {'c': 5, 'd': 6}

merged_overlap = dict_overlap_1 | dict_overlap_2 | dict_overlap_3
print(f"Merged with overlap: {merged_overlap}")
```
